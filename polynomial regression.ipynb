```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Dataset
X = np.array([-2, -1, 0, 1, 2]).reshape(-1, 1)
Y = np.array([9, 2, 1, 6, 17])

# Convert input into Polynomial Features (Degree 2)
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

# Create Linear Regression Model
model = LinearRegression()

# Train the model
model.fit(X_poly, Y)

# Predict output
Y_pred = model.predict(X_poly)

# Find coefficients
a = model.coef_[2]
b = model.coef_[1]
c = model.intercept_

# Display polynomial equation
print("Polynomial Regression Equation:")

print("y = {:.4f}x^2 + {:.4f}x + {:.4f}".format(a, b, c))

# Calculate Train MSE
train_mse = mean_squared_error(Y, Y_pred)

print("\nTrain-MSE:", round(train_mse, 4))

# Display actual and predicted values
print("\nX\tActual Y\tPredicted Y")

for i in range(len(X)):
    print(X[i][0], "\t", Y[i], "\t\t",
          round(Y_pred[i], 4))

# Plot the graph
X_curve = np.linspace(-2, 2, 100).reshape(-1, 1)

X_curve_poly = poly.transform(X_curve)

Y_curve = model.predict(X_curve_poly)

plt.scatter(X, Y, color='blue',
            label='Actual Data')

plt.plot(X_curve, Y_curve, color='red',
         label='Polynomial Regression Curve')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Polynomial Linear Regression (Degree 2)")

plt.legend()
plt.grid(True)
plt.show()
```
